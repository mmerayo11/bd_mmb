```python
import boto3

try:
    s3 = boto3.client('s3')
    buckets = s3.list_buckets()
    print("¡Conexión exitosa!")
    print(f"Tienes {len(buckets['Buckets'])} buckets en tu cuenta.")
except Exception as e:
    print("Error en la conexión. Revisa tus credenciales.")
    print(e)
```

```python
for bucket in buckets['Buckets']:
    nombre = bucket['Name']
    print (f"Nombre del bucket: {nombre}")
```

```python
s3.create_bucket(Bucket='mi-nombre')
```

```python
s3.delete_bucket(Bucket='mi-nombre')
```
