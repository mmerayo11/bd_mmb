## Ingesta Excel

```python
!pip install openpyxl
```

```python
df_ex = pd.read_excel(
    "ventas_sur.xlsx",
    sheet_name=None,
    header=0,
    usecols='A:E',
    names = ['id', 'date', 'product', 'quantity', 'unit_price'],
    parse_dates=['date'],
    dtype={'product' : 'string'},
    engine='openpyxl')
df_ex
```

```python
df2= pd.concat(
    df_ex.values(),
    ignore_index=True)
df2
```

## Ingesta Json
